```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
from PIL import Image
```

```python
data = pd.read_csv("data.csv")
data = data[["condition", "arm", "site", "delay_k", "efference_length", "noise_std", "seed_pair",
             "reward", "reward_1b", "actual_step", "steps_to_900"]]
```

```python
baseline_data = data[data.delay_k==0]
plt.figure(figsize=(5, 3), dpi=300)
sns.lineplot(data=baseline_data, x="noise_std", y="reward", errorbar="sd")
plt.ylim(0, 1000)
sns.despine()
plt.ylabel("Reward")
plt.xlabel("Size of additative gaussian action noise ($\sigma$)")
plt.title("Undelayed baseline")
im = Image.open(f"../three-arm-overview/illustrations/walker.png")
im = matplotlib.offsetbox.OffsetImage(im, zoom=0.12)
plt.gca().add_artist(matplotlib.offsetbox.AnnotationBbox(im, (0.4, 525), frameon=False))
plt.text(0.4, 150, "Task: Walker walk\nDelay: 0", style="italic", ha="center")
```

```python
plt.figure(figsize=(4, 2), dpi=300)
xx = np.linspace(-1,1,200)
yy = np.exp(-xx*xx/(0.5*0.5*2))
plt.plot(xx, yy, 'k')
plt.ylim(0, 1.01)
plt.yticks([])
plt.ylabel("pdf")
plt.xlabel("Added noise to action $\in$ [-1, 1]")
#plt.plot([-3.1, -1.1], [0.5, 0.5], 'k')
sns.despine(left=False)
```

```python
plt.figure(figsize=(6, 4), dpi=300)
d = data[data.noise_std==0.0]
sns.lineplot(data=d, x="delay_k", y="reward", hue="condition", hue_order=["WalkerWalk_no_copy", "WalkerWalk_copy"], legend=False)
d = data[data.noise_std==0.5]
sns.lineplot(data=d, x="delay_k", y="reward", hue="condition", palette=["C3", "C4", "C5"], legend=False)

plt.ylim(0, 1000)
sns.despine()
plt.ylabel("Reward")
plt.xlabel("Delay (25ms-steps)")
im = Image.open(f"../three-arm-overview/illustrations/walker.png")
im = matplotlib.offsetbox.OffsetImage(im, zoom=0.12)
plt.gca().add_artist(matplotlib.offsetbox.AnnotationBbox(im, (4, 325), frameon=False))
plt.text(4, 100, "Task: Walker walk", style="italic", ha="center")
plt.text(25.5, 425, "No noise + Action buffer", color="C1")
plt.text(25.5, 225, "No noise", color="C0")
plt.text(25.5, 175, "Predictable noise ($\sigma=0.5$) + Action buffer", color="C5")
plt.text(25.5, 125, "Noise ($\sigma=0.5$) + Action buffer", color="C4")
plt.text(25.5, 75, "Noise ($\sigma=0.5$)", color="C3")
```

```python
plt.figure(figsize=(6, 4), dpi=300)
d = data[data.noise_std==0.0]
sns.lineplot(data=d, x="delay_k", y="steps_to_900", hue="condition", hue_order=["WalkerWalk_no_copy", "WalkerWalk_copy"], legend=False)
d = data[data.noise_std==0.5]
sns.lineplot(data=d, x="delay_k", y="steps_to_900", hue="condition", palette=["C3", "C4", "C5"], legend=False)

plt.ylim(0, 1e9)
sns.despine()
plt.ylabel("Steps to reach 900 reward")
plt.xlabel("Delay (25ms-steps)")
im = Image.open(f"../three-arm-overview/illustrations/walker.png")
im = matplotlib.offsetbox.OffsetImage(im, zoom=0.12)
plt.gca().add_artist(matplotlib.offsetbox.AnnotationBbox(im, (16, 3.5e8), frameon=False))
plt.text(16, 1e8, "Task: Walker walk", style="italic", ha="center")
plt.yticks(np.arange(6)*2e8, [0]+[f"{200*i}M" for i in range(1,5)]+["1B"])
plt.plot([14.7, 15.7], [9.2e8]*2, color="C1")
plt.plot([14.7, 15.7], [8.7e8]*2, color="C0")
plt.plot([14.7, 15.7], [8.2e8]*2, color="C5")
plt.plot([14.7, 15.7], [7.7e8]*2, color="C4")
plt.plot([14.7, 15.7], [7.2e8]*2, color="C3")
plt.text(16, 9e8, "No noise + Action buffer", color="C1")
plt.text(16, 8.5e8, "No noise", color="C0")
plt.text(16, 8e8, "Predictable noise ($\sigma=0.5$) + Action buffer", color="C5")
plt.text(16, 7.5e8, "Noise ($\sigma=0.5$) + Action buffer", color="C4")
plt.text(16, 7e8, "Noise ($\sigma=0.5$)", color="C3")
```
