```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
from PIL import Image
```

```python
data = pd.read_csv("data.csv")
data = data[["condition", "critic", "delay", "efference_length", "seed",
             "reward_480", "actual_step", "steps_to_900"]]
```

```python
data.condition.unique()
```

```python
fig, axs = plt.subplots(1,2,sharey=True, figsize=(10, 3))
axs[0].plot(data[(data.condition == "privileged_efference") | (data.condition == "privileged_undelayed")].groupby("delay").reward_480.mean(), label="Privileged critic", color="")
axs[0].plot(data[(data.condition == "delayed_efference") | (data.condition == "delayed_undelayed")].groupby("delay").reward_480.mean(), label="Non-privileged critic")
axs[0].set_title("With efference copy")

axs[1].plot(data[(data.condition == "privileged_no_efference") | (data.condition == "privileged_undelayed")].groupby("delay").reward_480.mean(), label="Privileged critic")
axs[1].plot(data[(data.condition == "delayed_no_efference") | (data.condition == "delayed_undelayed")].groupby("delay").reward_480.mean(), label="Non-privileged critic")
axs[1].set_title("Without an efference copy")

for i in range(2):
    axs[i].legend()
    axs[i].set_xlabel("Delay (steps)")
axs[0].set_ylabel("Reward after 480M steps")
sns.despine()
```
